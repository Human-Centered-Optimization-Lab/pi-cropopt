```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import sys
import pickle

year = 2000
file_name = "/rdata/ian/pico/paperRuns/global_pf.pkl"

tab = pd.read_pickle(file_name)


#selection_mask = np.logical_and(np.logical_and(tab['run'] == 0, tab['DM_range'] == "25to30"), tab['gen'] == 200)
selection_mask = np.ones((tab.shape[0],1))
selection_mask[:] = True
print(selection_mask)

x = tab[selection_mask]['irr_total']
y = tab[selection_mask]['yield']

fig = plt.figure()
ax = fig.add_subplot()

ax.scatter(x, y)

ax.set_xlabel('Total irrigation')
ax.set_ylabel('Yield')

plt.title(year)
plt.show()
```
