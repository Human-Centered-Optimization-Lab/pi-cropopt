```python
import matplotlib.pyplot as plt
from pymoo.util.nds.non_dominated_sorting import NonDominatedSorting
import numpy as np
import pandas as pd
import sys
import pickle

year = 2000
file_name = r"/rdata/ian/pico/paperRuns/global_run_table.pkl"

tab = pd.read_pickle(file_name)
```

```python
selection_masks = [
    np.all([tab['algorithm'] == "pinsga2", tab['run'] == 0, tab['DM_range'] == "20to30", tab['gen'] == 200], axis=0),
    np.all([tab['algorithm'] == "nsga2", tab['run'] == 0,  tab['gen'] == 50], axis=0)]

labels = ["PI-NSGA-II", "NSGA-II"]

dfs = []
facecolors = ['none', 'green', 'purple', 'orange']
edgecolors = ['black', 'green', 'purple', 'orange']
markers = ['s', 'o', 'o', 'o']
```

```python
for (m, mask) in enumerate(selection_masks):
    dfs.append(tab.loc[mask, ['irr_total', 'yield']])

```

```python
for (d, df) in enumerate(dfs):

    # Perform non-dominated sorting
    nds = NonDominatedSorting()

    minimize_pop = df
    minimize_pop["yield"] = minimize_pop["yield"] * -1

    #print(minimize_pop)
    
    fronts = nds.do(minimize_pop.values, only_non_dominated_front=True)

    # Get the Pareto front
    pf = df.iloc[fronts,:]

    # Plot the data (first column is f1, second column is f2)
    plt.scatter(pf.iloc[:,0], pf.iloc[:,1]*-1,
                label=labels[d],
                facecolors=facecolors[d],
                edgecolors=edgecolors[d],
                marker=markers[d])



plt.xlabel("Irrigation (mm)")
plt.ylabel("Yield (kg/ha)")
plt.legend()
```
